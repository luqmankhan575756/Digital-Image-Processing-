Q12:  Image Morphing

```python
import os
import matplotlib.pyplot as plt
from PIL import Image

right = Image.open(r"D:\6th semester\DIP\Iright.ppm").convert("RGB")
left = Image.open(r"D:\6th semester\DIP\Ileft.ppm").convert("RGB")

if right.size != left.size:
    raise ValueError("Both images must have the same dimensions.")

os.makedirs("Morphing_Images", exist_ok=True)

for i in range(11):
    w = i / 10

    output_image = Image.blend(left, right, w)

    output_name = f"Morphing_Images/Morph_{i:02d}.ppm"

    output_image.save(output_name)

    plt.figure(figsize=(6, 5))
    plt.imshow(output_image)
    plt.title(f"Image Morphing: w = {w:.1f}")
    plt.axis("off")
    plt.show()

    print("Saved:", output_name)
```
