```python

import os

from PIL import Image
import matplotlib.pyplot as plt

images = [r"D:\6th semester\DIP\Iambient.ppm", r"D:\6th semester\DIP\Ileft.ppm", r"D:\6th semester\DIP\Iright.ppm"]

for filename in images:
    
    img = Image.open(filename).convert("RGB")


    R, G, B = img.split()


    zero = Image.new("L", img.size, 0)

    red_image = Image.merge("RGB", (R, zero, zero))


    output_name = "R_" + os.path.splitext(filename)[0] + ".png"

    # Display output image
    plt.figure(figsize=(6, 5))
    plt.imshow(red_image)
    plt.title("Red Layer: " + filename)
    plt.axis("off")
    plt.show()

    print("Saved:", output_name)
```
