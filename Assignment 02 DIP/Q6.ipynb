Q6: Keep G-Layer and B-Layer (Make R-Layer Zero)

```python
import os
import matplotlib.pyplot as plt
from PIL import Image

images = [r"D:\6th semester\DIP\Iambient.ppm", r"D:\6th semester\DIP\Ileft.ppm", r"D:\6th semester\DIP\Iright.ppm"]

for filename in images:
    img = Image.open(filename).convert("RGB")

    R, G, B = img.split()

    zero = Image.new("L", img.size, 0)

    output_image = Image.merge("RGB", (zero, G, B))

    output_name = "GB_" + os.path.splitext(os.path.basename(filename))[0] + ".png"

    output_image.save(output_name)

    plt.figure(figsize=(6, 5))
    plt.imshow(output_image)
    plt.title("Green and Blue Layers: " + os.path.basename(filename))
    plt.axis("off")
    plt.show()

    print("Saved:", output_name)
```
