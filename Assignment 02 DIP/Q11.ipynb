Q11) Generate redLeftGreenRightContribution.ppm

```python
import matplotlib.pyplot as plt
from PIL import Image

left = Image.open("LeftContribution.ppm").convert("RGB")
right = Image.open("RightContribution.ppm").convert("RGB")

R_left, G_left, B_left = left.split()
R_right, G_right, B_right = right.split()

zero = Image.new("L", left.size, 0)

output_image = Image.merge("RGB", (R_left, G_right, zero))

output_image.save("redLeftGreenRightContribution.ppm")

plt.figure(figsize=(6, 5))
plt.imshow(output_image)
plt.title("Red Left Green Right Contribution")
plt.axis("off")
plt.show()

print("Saved: redLeftGreenRightContribution.ppm")
```
