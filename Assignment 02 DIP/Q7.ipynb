Q7: Inverse of an Input Image

```python
import os
import matplotlib.pyplot as plt
from PIL import Image, ImageChops

images = [r"D:\6th semester\DIP\Iambient.ppm", r"D:\6th semester\DIP\Ileft.ppm", r"D:\6th semester\DIP\Iright.ppm"]


for filename in images:
    img = Image.open(filename).convert("RGB")

    inverse_image = ImageChops.invert(img)

    output_name = "Inverse_" + os.path.splitext(os.path.basename(filename))[0] + ".png"

    inverse_image.save(output_name)

    plt.figure(figsize=(6, 5))
    plt.imshow(inverse_image)
    plt.title("Inverse Image: " + os.path.basename(filename))
    plt.axis("off")
    plt.show()

    print("Saved:", output_name)
```
