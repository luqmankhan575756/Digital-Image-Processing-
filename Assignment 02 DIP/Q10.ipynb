Q10) Generate RightContribution.ppm

```python
import matplotlib.pyplot as plt
from PIL import Image, ImageChops

right = Image.open(r"D:\6th semester\DIP\Iright.ppm").convert("RGB")
ambient = Image.open(r"D:\6th semester\DIP\Iambient.ppm").convert("RGB")

right_contribution = ImageChops.subtract(right, ambient)

right_contribution.save("RightContribution.ppm")

plt.figure(figsize=(6, 5))
plt.imshow(right_contribution)
plt.title("Right Contribution")
plt.axis("off")
plt.show()

print("Saved: RightContribution.ppm")
```
