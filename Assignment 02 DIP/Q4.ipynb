Q4 Keep R-Layer and G-Layer (Make B-Layer Zero)

```python
import os
import matplotlib.pyplot as plt
from PIL import Image

images = [r"D:\6th semester\DIP\Iambient.ppm", r"D:\6th semester\DIP\Ileft.ppm", r"D:\6th semester\DIP\Iright.ppm"]

for filename in images:
    img = Image.open(filename).convert("RGB")

    R, G, B = img.split()

    zero = Image.new("L", img.size, 0)

    output_image = Image.merge("RGB", (R, G, zero))

    output_name = "RG_" + os.path.splitext(os.path.basename(filename))[0] + ".png"

    output_image.save(output_name)

    plt.figure(figsize=(6, 5))
    plt.imshow(output_image)
    plt.title("Red and Green Layers: " + os.path.basename(filename))
    plt.axis("off")
    plt.show()

    print("Saved:", output_name)
```
