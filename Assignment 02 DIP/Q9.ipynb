Q9 Generate LeftContribution.ppm

```python
import matplotlib.pyplot as plt
from PIL import Image, ImageChops

left = Image.open(r"D:\6th semester\DIP\Ileft.ppm").convert("RGB")
ambient = Image.open(r"D:\6th semester\DIP\Iambient.ppm").convert("RGB")

left_contribution = ImageChops.subtract(left, ambient)

left_contribution.save("LeftContribution.ppm")

plt.figure(figsize=(6, 5))
plt.imshow(left_contribution)
plt.title("Left Contribution")
plt.axis("off")
plt.show()

print("Saved: LeftContribution.ppm")
```
