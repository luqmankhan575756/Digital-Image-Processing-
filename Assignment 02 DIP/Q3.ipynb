```python

import os
import matplotlib.pyplot as plt
from PIL import Image

images = [r"D:\6th semester\DIP\Iambient.ppm", r"D:\6th semester\DIP\Ileft.ppm", r"D:\6th semester\DIP\Iright.ppm"]

for filename in images:
    img = Image.open(filename).convert("RGB")

    R, G, B = img.split()

    zero = Image.new("L", img.size, 0)

    blue_image = Image.merge("RGB", (zero, zero, B))

    output_name = "B_" + os.path.splitext(filename)[0] + ".png"

    plt.figure(figsize=(6, 5))
    plt.imshow(blue_image)
    plt.title("Blue Layer: " + filename)
    plt.axis("off")
    plt.show()

    print("Saved:", output_name)
```
