Q8 Convert an RGB Image into a Grayscale Image

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

images = [r"D:\6th semester\DIP\Iambient.ppm", r"D:\6th semester\DIP\Ileft.ppm", r"D:\6th semester\DIP\Iright.ppm"]



for filename in images:
    img = Image.open(filename).convert("RGB")

    R, G, B = img.split()

    R = np.array(R, dtype=float)
    G = np.array(G, dtype=float)
    B = np.array(B, dtype=float)

    gray_array = (0.3 * R + 0.59 * G + 0.11 * B).astype(np.uint8)

    gray_image = Image.fromarray(gray_array)

    output_name = "Gray_" + os.path.splitext(os.path.basename(filename))[0] + ".png"

    gray_image.save(output_name)

    plt.figure(figsize=(6, 5))
    plt.imshow(gray_image, cmap="gray")
    plt.title("Grayscale Image: " + os.path.basename(filename))
    plt.axis("off")
    plt.show()

    print("Saved:", output_name)
```
